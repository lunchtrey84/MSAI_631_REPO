```python
#MovieMatch Chatbot!

def chatbot():
    print("MovieMatch: Welcome! Type 'help' to see what I can do.")

    while True:
        user_input = input("You: ").lower().strip()

        if user_input in ["bye", "quit", "exit"]:
            print("MovieMatch: Enjoy your movie!")
            break

        elif user_input in ["help", "what can you do"]:
            print("MovieMatch: I can recommend action, comedy, horror, sci-fi, or drama movies.")

        elif "action" in user_input:
            print("MovieMatch: Try Mad Max: Fury Road.")

        elif "comedy" in user_input or "funny" in user_input:
            print("MovieMatch: Try The Big Lebowski.")

        elif "horror" in user_input or "scary" in user_input:
            print("MovieMatch: Try The Thing.")

        elif "sci-fi" in user_input or "science fiction" in user_input:
            print("MovieMatch: Try Blade Runner.")

        elif "drama" in user_input:
            print("MovieMatch: Try The Shawshank Redemption.")

        elif not user_input:
            print("MovieMatch: Please enter a question or movie preference.")

        else:
            print("MovieMatch: I don't understand that. Type 'help' to see my options.")

chatbot()
```
